```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_csv('student_sleep_screen_time_academic_performance.csv')


df.head()
```

```python
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8,6))

sns.regplot(data=df, x="sleep_quality_score", y="cgpa", scatter_kws={"alpha":0.5}, line_kws={"color":"red"})

plt.title("Relationship Between Sleep Quality And Cgpa", fontsize=14)
plt.xlabel("Sleep Quality", fontsize=12)
plt.ylabel("CGPA", fontsize=14)
```

```python
plt.figure(figsize=(8, 6))
sns.set_theme(style="whitegrid")

sns.boxplot(data=df, x='has_part_time_job', y='cgpa', palette='Set2')

plt.title('The Impact of Part-Time Jobs on Academic Performance (CGPA)', fontsize=12)
plt.xlabel('Has Part-Time Job?', fontsize=11)
plt.ylabel('CGPA', fontsize=11)

plt.show()
```
